```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Cấu hình hiển thị
pd.set_option('display.max_columns', None)
sns.set_theme(style="whitegrid")
```

```
# 1. Tải và Làm sạch Dữ liệu Tech Companies
tech_df = pd.read_csv('dataset/Top 50 US Tech Companies 2022 - 2023.csv')

# Kiểm tra và xử lý dữ liệu bị thiếu
print("Missing values in Tech Companies:\n", tech_df.isnull().sum())
tech_df = tech_df.dropna()

# Chuyển đổi các cột số bị lỗi định dạng (nếu có)
# Ví dụ: Employee Size sang dạng int
tech_df['Employee Size'] = pd.to_numeric(tech_df['Employee Size'], errors='coerce')

# 2. Tải và Làm sạch Dữ liệu Constituents Financials
fin_df = pd.read_csv('dataset/constituents-financials.csv')
print("\nFinancials Data Shape:", fin_df.shape)

# Điền các giá trị NaN bằng giá trị trung vị (median) cho các cột số
num_cols = fin_df.select_dtypes(include=np.number).columns
fin_df[num_cols] = fin_df[num_cols].fillna(fin_df[num_cols].median())

# 3. Tải Dữ liệu Scatter Data
scatter_df = pd.read_csv('dataset/scatter-data.csv')
print("\nScatter Data Shape:", scatter_df.shape)
scatter_df = scatter_df.dropna()

# 4. Tải và Làm sạch Dữ liệu S&P 500 Daily Update
# Lưu ý: File này rất lớn (158MB), ta có thể load thử một số cột cơ bản hoặc chunk
try:
    snp_df = pd.read_csv('dataset/SnP_daily_update.csv', nrows=5000) # Tạm load 5000 dòng để xử lý sơ bộ
    print("\nS&P 500 Sample Shape:", snp_df.shape)
except Exception as e:
    print("Error loading S&P 500:", e)

tech_df.head()
```

```
# Đổi tên cột để dễ ghép nối dữ liệu
fin_df.rename(columns={'Symbol': 'Stock Name'}, inplace=True)

# Merge dữ liệu Tech Companies với Financials để có thêm các chỉ số P/E, Dividend Yield,...
merged_df = pd.merge(tech_df, fin_df, on='Stock Name', how='inner')

print("Merged Data Shape:", merged_df.shape)
merged_df.head()
```
